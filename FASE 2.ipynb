```python
import matplotlib.pyplot as plt #para graficos
import numpy as np
import pandas as pd #para embellecer los graficos (lo uso cuando las funciones son muy variables, y los graficos)
```

## FASE 1.1 La funcion de Collatz (ESENCIAL PARA ESTA FASE)

Definimos una funcion que recibe un numero entero positivo y devuelve
la trayectoria completa hasta llegar a 1.

```python
def collatz(n):
    """Genera la trayectoria de Collatz partiendo desde n."""
    trayectoria = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        trayectoria.append(n)
    return trayectoria
```

sacar informacion de altitud y tiempo de vuelo


```python
def info_collatz(n):
    trayectoria = collatz(n)
    pasos = len(trayectoria) - 1
    maximo = max(trayectoria)
    return pasos, maximo
pasos, maximo = info_collatz(100) #se reemplaza el 27 por cualquier numero para ver sus datos

print("Tiempo de vuelo T(n):", pasos)
print("Altitud máxima A(n):", maximo)
```

## FASE 2.1 Calcular la media la mediana, la moda y la desviacion estndar del tiempo de vuelo

```python
rango = range(1, 100001)

maximos_lista = [max(collatz(n)) for n in rango]

y_bonita = pd.Series(maximos_lista).rolling(window=75, min_periods=1).mean()

plt.figure(figsize=(12, 5))

plt.plot(rango, y_bonita, color='red', label='altitud')

plt.xlabel("valor de n")
plt.ylabel("Altitud máxima A(n)")
plt.title("Altitud máxima en Collatz (1 hasta 100000)")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.yscale("log")
plt.xscale("log")
plt.show()

promedio = y_bonita.mean()
mediana = y_bonita.median()
moda = y_bonita.mode()[0]
desviacion_estandar = y_bonita.std()
print("RESULTADOS prom, med y mod ")
print(f"promedio:   {promedio:.2f}")
print(f"mediana: {mediana:.2f}")
print(f"moda:    {moda:.2f}")
print(f"desviacion estandar: {desviacion_estandar:.2f}")
```
